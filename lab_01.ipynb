## Индексирование и булев информационный поиск

Вариант 1: `random_state=101`, `N=500`, `min_len=2`, оператор `AND`.

```python
from pathlib import Path
from solution import load_documents, sample_documents, build_inverted_index, boolean_search
import time
```

```python
DATA = Path('../result.jsonl')  # при необходимости скорректируйте путь
docs_all = load_documents(DATA)
len(docs_all)
```

```python
docs = sample_documents(docs_all, n=500, random_state=101)
t0 = time.perf_counter()
index = build_inverted_index(docs, min_len=2)
build_time = time.perf_counter() - t0
len(index), build_time
```

```python
query = 'magic world'
t0 = time.perf_counter()
terms, ids = boolean_search(query, index, operator='AND', min_len=2)
search_time = time.perf_counter() - t0
terms, len(ids), search_time
```

```python
[(doc_id, docs[doc_id]['title']) for doc_id in ids[:10]]
```

### Вывод
Инвертированный индекс хранит соответствие между терминами и документами. AND-поиск пересекает posting lists, OR-поиск объединяет их. Поиск по готовому индексу выполняется значительно быстрее полного просмотра корпуса.
